## 4. Byte-Level BPE Tokenizer Loading and Verification
Initialization of the 25,000-vocabulary BPE tokenizer trained on the Indian legal corpus.


```python
#tokenizer from A1
import os
import json

BPE_FILENAME = "bpe_en_25k.json"
bpe_candidates = []

for root, dirs, files in os.walk("/kaggle/input"):
    if BPE_FILENAME in files:
        bpe_candidates.append(os.path.join(root, BPE_FILENAME))

if len(bpe_candidates) == 0:
    raise FileNotFoundError(
        f"{BPE_FILENAME} was not found inside /kaggle/input/. "
        "Make sure the BPE-25k dataset is attached."
    )

BPE_PATH = bpe_candidates[0]

print("Tokenizer found:", BPE_PATH)


# Load the vocabulary and learned BPE merges.
with open(BPE_PATH, "r", encoding="utf-8") as f:
    bpe_data = json.load(f)

bpe_itos = bpe_data["itos"]
bpe_merges = bpe_data["merges"]
bpe_vocab_size = bpe_data["vocab_size"]

# Make sure we have loaded the tokenizer we actually want.
assert bpe_vocab_size == 25_000
assert len(bpe_itos) == 25_000

print("Vocabulary size:", bpe_vocab_size)
print("Number of merge rules:", len(bpe_merges))


# Token -> ID mapping
bpe_stoi = {
    token: idx
    for idx, token in enumerate(bpe_itos)
}
merge_to_id = {}

for merge_index, merge in enumerate(bpe_merges):
    left_token, right_token = merge

    left_id = bpe_stoi[left_token]
    right_id = bpe_stoi[right_token]

    new_token_id = 261 + merge_index
    merge_to_id[(left_id, right_id)] = new_token_id

PAD_TOKEN_ID = bpe_stoi["<pad>"]
UNK_TOKEN_ID = bpe_stoi["<unk>"]
BOS_TOKEN_ID = bpe_stoi["<s>"]
EOS_TOKEN_ID = bpe_stoi["</s>"]
WORD_BOUNDARY_ID = bpe_stoi["<wb>"]

print("\nSpecial token IDs:")
print("PAD:", PAD_TOKEN_ID)
print("UNK:", UNK_TOKEN_ID)
print("BOS:", BOS_TOKEN_ID)
print("EOS:", EOS_TOKEN_ID)
print("WORD BOUNDARY:", WORD_BOUNDARY_ID)
```

```python
#Tokenizer Sanity Check
#Before using the tokenizer on the full legal corpus,let's check a small example and make sure encoding and decoding work correctly.

class ByteLevelBPE25k:
    def __init__(self, itos, stoi, merge_to_id, merges):
        self.itos = itos
        self.stoi = stoi
        self.merge_to_id = merge_to_id
        self.merges = merges
        self.vocab_size = len(itos)

        self.pad_token_id = stoi["<pad>"]
        self.unk_token_id = stoi["<unk>"]
        self.bos_token_id = stoi["<s>"]
        self.eos_token_id = stoi["</s>"]
        self.word_boundary_id = stoi["<wb>"]

    def _encode_word(self, word):
        # Start with the UTF-8 bytes of the word.
        sequence = [
            4 + byte
            for byte in word.encode("utf-8")
        ]

        # Our BPE representation keeps a word-boundary token.
        sequence.append(self.word_boundary_id)

        while len(sequence) > 1:
            best_position = None
            best_token_id = None

            for i in range(len(sequence) - 1):
                pair = (sequence[i], sequence[i + 1])
                new_token_id = self.merge_to_id.get(pair)

                if new_token_id is None:
                    continue

                if (
                    best_token_id is None
                    or new_token_id < best_token_id
                ):
                    best_token_id = new_token_id
                    best_position = i

            if best_position is None:
                break

            i = best_position
            sequence = (
                sequence[:i]
                + [best_token_id]
                + sequence[i + 2:]
            )

        return sequence

    def encode(self, text, add_bos=False, add_eos=False):
        # The original tokenizer was trained word-by-word.
        words = text.split()

        token_ids = []

        if add_bos:
            token_ids.append(self.bos_token_id)

        for word in words:
            token_ids.extend(
                self._encode_word(word)
            )

        if add_eos:
            token_ids.append(self.eos_token_id)

        return token_ids

    def _token_to_bytes(self, token_id):
        token = self.itos[token_id]

        # Raw byte token.
        if token.startswith("<0x") and token.endswith(">"):
            return bytes([int(token[3:-1], 16)])

        if token == "<wb>":
            return b" "

        if token in {"<pad>", "<unk>", "<s>", "</s>"}:
            return b""

        # Learned BPE token.
        if token.startswith("<bpe_"):
            merge_index = token_id - 261

            if 0 <= merge_index < len(self.merges):
                left_token, right_token = self.merges[merge_index]

                left_id = self.stoi[left_token]
                right_id = self.stoi[right_token]

                return (
                    self._token_to_bytes(left_id)
                    + self._token_to_bytes(right_id)
                )

        return b""

    def decode(self, token_ids):
        output = bytearray()

        for token_id in token_ids:
            if not 0 <= int(token_id) < self.vocab_size:
                continue

            output.extend(
                self._token_to_bytes(int(token_id))
            )

        return bytes(output).decode(
            "utf-8",
            errors="replace"
        )


# Creating the tokenizer.
tokenizer = ByteLevelBPE25k(
    itos=bpe_itos,
    stoi=bpe_stoi,
    merge_to_id=merge_to_id,
    merges=bpe_merges
)


# Test it on a small legal-style sentence.
sample_text = (
    "The Supreme Court of India has jurisdiction "
    "over constitutional matters."
)

sample_tokens = tokenizer.encode(sample_text)
decoded_text = tokenizer.decode(sample_tokens)

print("Original text:")
print(sample_text)
print("\nToken IDs:")
print(sample_tokens)
print("\nNumber of tokens:", len(sample_tokens))
print("\nDecoded text:")
print(decoded_text)
print("\nVocabulary size:", tokenizer.vocab_size)
assert tokenizer.vocab_size == 25_000
assert len(sample_tokens) > 0
print("\nTokenizer sanity check passed.")
```

## 6. Token Memory Mapping
Memory-mapped array setup for zero-copy streaming of token IDs during training.


```python
# 6. Create Train and Validation Token Cache
import os
import json
import random
import time
import numpy as np

CACHE_DIR = os.path.join(
    OUTPUT_DIR,
    "token_cache"
)

os.makedirs(CACHE_DIR, exist_ok=True)

TRAIN_TOKEN_PATH = os.path.join(
    CACHE_DIR,
    "train_tokens.bin"
)

VAL_TOKEN_PATH = os.path.join(
    CACHE_DIR,
    "val_tokens.bin"
)

TOKEN_CACHE_META = os.path.join(
    CACHE_DIR,
    "cache_metadata.json"
)

# Load the official training corpus
with open(train_file_path, "r", encoding="utf-8") as f:
    records = json.load(f)

print("Number of records:", len(records))

# Resume from an existing cache if possible

train_count = 0
val_count = 0
start_record = 0

if (
    os.path.exists(TRAIN_TOKEN_PATH)
    and os.path.exists(VAL_TOKEN_PATH)
    and os.path.exists(TOKEN_CACHE_META)
):
    with open(TOKEN_CACHE_META, "r", encoding="utf-8") as f:
        cache_meta = json.load(f)

    # Make sure this cache belongs to the same experiment.
    if (
        cache_meta.get("train_tokens") == TRAIN_TOKENS
        and cache_meta.get("val_tokens") == VAL_TOKENS
        and cache_meta.get("vocab_size") == tokenizer.vocab_size
        and cache_meta.get("seed") == SEED
    ):
        start_record = cache_meta["next_record"]
        train_count = cache_meta["train_count"]
        val_count = cache_meta["val_count"]

        print("Existing token cache found.")
        print("Resuming from record:", start_record)
        print("Train tokens already cached:", f"{train_count:,}")
        print("Validation tokens already cached:", f"{val_count:,}")

    else:
        print("Existing cache does not match the current settings.")
        print("Starting a fresh token cache.")

        os.remove(TRAIN_TOKEN_PATH)
        os.remove(VAL_TOKEN_PATH)
        os.remove(TOKEN_CACHE_META)

else:
    print("No existing token cache found.")
    print("Starting tokenization from the beginning.")

# Open cache files in append mode

train_file = open(
    TRAIN_TOKEN_PATH,
    "ab"
)

val_file = open(
    VAL_TOKEN_PATH,
    "ab"
)

# Tokenize the corpus
# We keep the same record-level random split used earlier:roughly 95% for training and 5% for validation.The seed and record index are saved, so the split stays fixed when the process is resumed.

random.seed(SEED)

# Recreate the same random decisions for records that were
# already processed before an interruption.
for _ in range(start_record):
    random.random()

start_time = time.time()

for record_index in range(start_record, len(records)):

    # Stop once both requested token budgets are reached.
    if (
        train_count >= TRAIN_TOKENS
        and val_count >= VAL_TOKENS
    ):
        break

    text = records[record_index]["text"]

    # Use our custom BPE-25k tokenizer.
    tokens = tokenizer.encode(
        text,
        add_bos=False,
        add_eos=False
    )

    # Decide whether this record belongs to validation.
    is_validation = random.random() < 0.05

    if is_validation and val_count < VAL_TOKENS:

        remaining = VAL_TOKENS - val_count
        tokens = tokens[:remaining]

        if tokens:
            token_array = np.asarray(
                tokens,
                dtype=np.uint32
            )

            val_file.write(
                token_array.tobytes()
            )

            val_count += len(token_array)

    elif not is_validation and train_count < TRAIN_TOKENS:

        remaining = TRAIN_TOKENS - train_count
        tokens = tokens[:remaining]

        if tokens:
            token_array = np.asarray(
                tokens,
                dtype=np.uint32
            )

            train_file.write(
                token_array.tobytes()
            )

            train_count += len(token_array)


    # Save progress periodically so an interruption does notforce us to start tokenization from zero.
    if (
        (record_index + 1) % 100 == 0
        or train_count >= TRAIN_TOKENS
        or val_count >= VAL_TOKENS
    ):

        train_file.flush()
        val_file.flush()

        cache_meta = {
            "train_tokens": TRAIN_TOKENS,
            "val_tokens": VAL_TOKENS,
            "train_count": train_count,
            "val_count": val_count,
            "next_record": record_index + 1,
            "vocab_size": tokenizer.vocab_size,
            "seed": SEED
        }

        with open(
            TOKEN_CACHE_META,
            "w",
            encoding="utf-8"
        ) as f:
            json.dump(
                cache_meta,
                f,
                indent=2
            )

    # Print progress every 100 records.
    if (record_index + 1) % 100 == 0:

        elapsed = time.time() - start_time

        print(
            f"Record {record_index + 1:,}/{len(records):,} | "
            f"Train: {train_count:,}/{TRAIN_TOKENS:,} | "
            f"Val: {val_count:,}/{VAL_TOKENS:,} | "
            f"Time: {elapsed / 60:.1f} min"
        )


train_file.flush()
val_file.flush()

train_file.close()
val_file.close()

# Final checks

assert train_count == TRAIN_TOKENS, (
    f"Expected {TRAIN_TOKENS:,} train tokens, "
    f"got {train_count:,}"
)

assert val_count == VAL_TOKENS, (
    f"Expected {VAL_TOKENS:,} validation tokens, "
    f"got {val_count:,}"
)

print("\nTokenization complete.")
print("Train tokens:", f"{train_count:,}")
print("Validation tokens:", f"{val_count:,}")
print("Train cache:", TRAIN_TOKEN_PATH)
print("Validation cache:", VAL_TOKEN_PATH)
```

## 7. Token Block Windowing
Segmentation of token sequences into fixed context blocks of length 384.


```python
# Create Dataset From Cached Tokens

import numpy as np
import torch
from torch.utils.data import Dataset

class LegalTokenDataset(Dataset):
    def __init__(
        self,
        token_path,
        num_tokens,
        block_size,
        stride=1
    ):
        self.token_path = token_path
        self.num_tokens = num_tokens
        self.block_size = block_size
        self.stride = stride

        # The file stays on disk and is accessed only when
        # a particular training example is requested.
        self.tokens = np.memmap(
            token_path,
            dtype=np.uint32,
            mode="r",
            shape=(num_tokens,)
        )

    def __len__(self):
        return 1 + (
            self.num_tokens
            - self.block_size
            - 1
        ) // self.stride

    def __getitem__(self, index):
        start = index * self.stride
        end = start + self.block_size + 1

        tokens = self.tokens[start:end].astype(
            np.int64
        )

        input_ids = torch.from_numpy(
            tokens[:-1]
        )

        labels = torch.from_numpy(
            tokens[1:]
        )

        return {
            "input_ids": input_ids,
            "labels": labels
        }

train_dataset = LegalTokenDataset(
    token_path=TRAIN_TOKEN_PATH,
    num_tokens=TRAIN_TOKENS,
    block_size=TRAIN_BLOCK_SIZE,
    stride=1
)

val_dataset = LegalTokenDataset(
    token_path=VAL_TOKEN_PATH,
    num_tokens=VAL_TOKENS,
    block_size=TRAIN_BLOCK_SIZE,
    stride=TRAIN_BLOCK_SIZE
)

print("Train examples:", f"{len(train_dataset):,}")
print("Validation examples:", f"{len(val_dataset):,}")
print("Sequence length:", TRAIN_BLOCK_SIZE)
```

```python
# 9. V2 Lightweight Hybrid Decoder-Only Transformer
#
# V2 preserves the successful V1 components:
# - RMSNorm
# - RoPE
# - Grouped Query Attention (GQA)
# - Causal self-attention
# - Standard GELU FFN
# - Residual connections
# - KV cache
#
# New architectural component:
# - Hybrid Global + Local Causal Attention
#
# Global layers:
#   Full causal attention over the complete sequence
#
# Local layers:
#   Causal attention restricted to a fixed local window
#
# The hybrid design is motivated by the course discussion of
# sparse/local attention and hybrid models that interleave
# full and sparse attention layers.

import math
import torch
import torch.nn as nn
import torch.nn.functional as F


# ---------------------------------------------------------
# RMSNorm
# ---------------------------------------------------------

class RMSNorm(nn.Module):
    def __init__(self, dim, eps=1e-6):
        super().__init__()

        self.weight = nn.Parameter(
            torch.ones(dim)
        )
        self.eps = eps

    def forward(self, x):
        rms = x.pow(2).mean(
            dim=-1,
            keepdim=True
        )

        x = x * torch.rsqrt(
            rms + self.eps
        )

        return self.weight * x


# ---------------------------------------------------------
# Rotary Positional Embeddings
# ---------------------------------------------------------

class RotaryEmbedding(nn.Module):
    def __init__(
        self,
        head_dim,
        max_seq_len,
        base=10000
    ):
        super().__init__()

        self.head_dim = head_dim
        self.max_seq_len = max_seq_len

        inv_freq = 1.0 / (
            base ** (
                torch.arange(
                    0,
                    head_dim,
                    2
                ).float()
                / head_dim
            )
        )

        self.register_buffer(
            "inv_freq",
            inv_freq,
            persistent=False
        )

        self._build_cache(max_seq_len)

    def _build_cache(self, seq_len):
        positions = torch.arange(
            seq_len,
            device=self.inv_freq.device
        ).float()

        freqs = torch.outer(
            positions,
            self.inv_freq
        )

        self.register_buffer(
            "cos_cached",
            freqs.cos(),
            persistent=False
        )

        self.register_buffer(
            "sin_cached",
            freqs.sin(),
            persistent=False
        )

    def forward(
        self,
        q,
        k,
        position_offset=0
    ):
        seq_len = q.size(-2)

        required_len = (
            position_offset + seq_len
        )

        if required_len > self.max_seq_len:
            self._build_cache(required_len)
            self.max_seq_len = required_len

        cos = self.cos_cached[
            position_offset:
            position_offset + seq_len
        ]

        sin = self.sin_cached[
            position_offset:
            position_offset + seq_len
        ]

        cos = cos.unsqueeze(0).unsqueeze(0)
        sin = sin.unsqueeze(0).unsqueeze(0)

        q = self._apply_rotary(
            q,
            cos,
            sin
        )

        k = self._apply_rotary(
            k,
            cos,
            sin
        )

        return q, k

    @staticmethod
    def _apply_rotary(x, cos, sin):
        x1 = x[..., ::2]
        x2 = x[..., 1::2]

        rotated = torch.stack(
            (-x2, x1),
            dim=-1
        ).flatten(-2)

        cos_full = torch.repeat_interleave(
            cos,
            2,
            dim=-1
        )

        sin_full = torch.repeat_interleave(
            sin,
            2,
            dim=-1
        )

        return (
            x * cos_full
            + rotated * sin_full
        )


# ---------------------------------------------------------
# Hybrid Grouped Query Attention
# ---------------------------------------------------------

class GQAAttention(nn.Module):
    def __init__(
        self,
        n_embd,
        n_head,
        n_kv_head,
        max_seq_len,
        attention_type="global",
        local_window=256,
        dropout=0.0
    ):
        super().__init__()

        assert n_embd % n_head == 0
        assert n_head % n_kv_head == 0

        assert attention_type in {
            "global",
            "local"
        }

        assert local_window > 0

        self.n_head = n_head
        self.n_kv_head = n_kv_head
        self.head_dim = n_embd // n_head

        self.kv_repeat = (
            n_head // n_kv_head
        )

        self.attention_type = attention_type
        self.local_window = local_window

        self.q_proj = nn.Linear(
            n_embd,
            n_head * self.head_dim,
            bias=False
        )

        self.k_proj = nn.Linear(
            n_embd,
            n_kv_head * self.head_dim,
            bias=False
        )

        self.v_proj = nn.Linear(
            n_embd,
            n_kv_head * self.head_dim,
            bias=False
        )

        self.out_proj = nn.Linear(
            n_embd,
            n_embd,
            bias=False
        )

        self.rope = RotaryEmbedding(
            head_dim=self.head_dim,
            max_seq_len=max_seq_len
        )

        self.dropout = dropout

    def _build_attention_mask(
        self,
        query_length,
        key_length,
        query_position_offset,
        device
    ):
        """
        Build a boolean attention mask.

        True  = attention is allowed
        False = attention is blocked

        For global attention:
            causal mask

        For local attention:
            causal + fixed sliding window
        """

        query_positions = (
            query_position_offset
            + torch.arange(
                query_length,
                device=device
            )
        )

        key_positions = torch.arange(
            key_length,
            device=device
        )

        # Causal condition:
        # query at position q can only see keys <= q.
        causal_mask = (
            key_positions.unsqueeze(0)
            <= query_positions.unsqueeze(1)
        )

        if self.attention_type == "global":
            return causal_mask

        # Local causal attention:
        # q can see at most the previous
        # local_window - 1 tokens plus itself.
        local_mask = (
            key_positions.unsqueeze(0)
            >= (
                query_positions.unsqueeze(1)
                - self.local_window
                + 1
            )
        )

        return causal_mask & local_mask

    def forward(
        self,
        x,
        past_key_value=None,
        use_cache=False
    ):
        B, T, C = x.shape

        # -------------------------------------------------
        # Q, K, V projections
        # -------------------------------------------------

        q = self.q_proj(x)
        k = self.k_proj(x)
        v = self.v_proj(x)

        q = q.view(
            B,
            T,
            self.n_head,
            self.head_dim
        ).transpose(1, 2)

        k = k.view(
            B,
            T,
            self.n_kv_head,
            self.head_dim
        ).transpose(1, 2)

        v = v.view(
            B,
            T,
            self.n_kv_head,
            self.head_dim
        ).transpose(1, 2)

        # -------------------------------------------------
        # KV cache information
        # -------------------------------------------------

        past_len = 0

        if past_key_value is not None:
            past_k, past_v = past_key_value
            past_len = past_k.size(-2)

        # -------------------------------------------------
        # RoPE
        # -------------------------------------------------

        q, k = self.rope(
            q,
            k,
            position_offset=past_len
        )

        # -------------------------------------------------
        # Append new K/V to cache
        # -------------------------------------------------

        if past_key_value is not None:
            k = torch.cat(
                [past_k, k],
                dim=-2
            )

            v = torch.cat(
                [past_v, v],
                dim=-2
            )

        present = None

        if use_cache:
            present = (k, v)

        # -------------------------------------------------
        # GQA
        # -------------------------------------------------

        if self.kv_repeat > 1:
            k = k.repeat_interleave(
                self.kv_repeat,
                dim=1
            )

            v = v.repeat_interleave(
                self.kv_repeat,
                dim=1
            )

        # -------------------------------------------------
        # Hybrid attention mask
        # -------------------------------------------------

        key_length = k.size(-2)

        attn_mask = self._build_attention_mask(
            query_length=T,
            key_length=key_length,
            query_position_offset=past_len,
            device=x.device
        )

        # -------------------------------------------------
        # Scaled dot-product attention
        # -------------------------------------------------

        y = F.scaled_dot_product_attention(
            q,
            k,
            v,
            attn_mask=attn_mask,
            dropout_p=(
                self.dropout
                if self.training
                else 0.0
            ),
            is_causal=False
        )

        # -------------------------------------------------
        # Output projection
        # -------------------------------------------------

        y = y.transpose(
            1,
            2
        ).contiguous()

        y = y.view(
            B,
            T,
            C
        )

        y = self.out_proj(y)

        return y, present


# ---------------------------------------------------------
# Standard GELU FFN
# ---------------------------------------------------------

class FeedForward(nn.Module):
    def __init__(
        self,
        n_embd,
        multiplier=4,
        dropout=0.0
    ):
        super().__init__()

        hidden_dim = int(
            n_embd * multiplier
        )

        self.fc1 = nn.Linear(
            n_embd,
            hidden_dim
        )

        self.fc2 = nn.Linear(
            hidden_dim,
            n_embd
        )

        self.dropout = nn.Dropout(
            dropout
        )

    def forward(self, x):
        x = self.fc1(x)
        x = F.gelu(x)
        x = self.fc2(x)
        x = self.dropout(x)

        return x


# ---------------------------------------------------------
# Hybrid Transformer Block
# ---------------------------------------------------------

class TransformerBlock(nn.Module):
    def __init__(
        self,
        n_embd,
        n_head,
        n_kv_head,
        max_seq_len,
        ffn_multiplier,
        dropout,
        layer_index,
        global_attention_layers,
        local_window
    ):
        super().__init__()

        self.layer_index = layer_index

        if layer_index in global_attention_layers:
            attention_type = "global"
        else:
            attention_type = "local"

        self.attention_type = attention_type

        self.norm1 = RMSNorm(n_embd)

        self.attention = GQAAttention(
            n_embd=n_embd,
            n_head=n_head,
            n_kv_head=n_kv_head,
            max_seq_len=max_seq_len,
            attention_type=attention_type,
            local_window=local_window,
            dropout=dropout
        )

        self.norm2 = RMSNorm(n_embd)

        self.ffn = FeedForward(
            n_embd=n_embd,
            multiplier=ffn_multiplier,
            dropout=dropout
        )

    def forward(
        self,
        x,
        past_key_value=None,
        use_cache=False
    ):
        # Pre-norm attention + residual.
        attn_output, present = self.attention(
            self.norm1(x),
            past_key_value=past_key_value,
            use_cache=use_cache
        )

        x = x + attn_output

        # Pre-norm FFN + residual.
        x = x + self.ffn(
            self.norm2(x)
        )

        return x, present


# ---------------------------------------------------------
# V2 Lightweight Hybrid Legal Language Model
# ---------------------------------------------------------

class HybridLegalLM(nn.Module):
    def __init__(
        self,
        vocab_size,
        n_embd,
        n_layer,
        n_head,
        n_kv_head,
        max_seq_len,
        ffn_multiplier,
        dropout
    ):
        super().__init__()

        self.vocab_size = vocab_size
        self.n_embd = n_embd

        self.token_embedding = nn.Embedding(
            vocab_size,
            n_embd
        )

        self.blocks = nn.ModuleList([
            TransformerBlock(
                n_embd=n_embd,
                n_head=n_head,
                n_kv_head=n_kv_head,
                max_seq_len=max_seq_len,
                ffn_multiplier=ffn_multiplier,
                dropout=dropout,
                layer_index=layer_index,
                global_attention_layers=GLOBAL_ATTENTION_LAYERS,
                local_window=LOCAL_ATTN_WINDOW
            )
            for layer_index in range(n_layer)
        ])

        self.final_norm = RMSNorm(
            n_embd
        )

        # Weight tying retained from V1.
        self.lm_head = nn.Linear(
            n_embd,
            vocab_size,
            bias=False
        )

        self.lm_head.weight = (
            self.token_embedding.weight
        )

        self.apply(self._init_weights)

    def _init_weights(self, module):
        if isinstance(module, nn.Linear):
            nn.init.normal_(
                module.weight,
                mean=0.0,
                std=0.02
            )

            if module.bias is not None:
                nn.init.zeros_(
                    module.bias
                )

        elif isinstance(module, nn.Embedding):
            nn.init.normal_(
                module.weight,
                mean=0.0,
                std=0.02
            )

    def forward(
        self,
        input_ids,
        past_key_values=None,
        use_cache=False
    ):
        x = self.token_embedding(
            input_ids
        )

        presents = [] if use_cache else None

        for layer_index, block in enumerate(
            self.blocks
        ):
            past = None

            if past_key_values is not None:
                past = past_key_values[
                    layer_index
                ]

            x, present = block(
                x,
                past_key_value=past,
                use_cache=use_cache
            )

            if use_cache:
                presents.append(
                    present
                )

        x = self.final_norm(x)

        logits = self.lm_head(x)

        return logits, presents
```

```python
# Hybrid Local/Global Attention Mask Test
#
# Verify:
# 1. Global layers allow full causal history.
# 2. Local layers only attend within LOCAL_ATTN_WINDOW.
# 3. Local attention includes the current token.
# 4. Cached decoding uses the same absolute-position rule.

model.eval()

global_layer_idx = GLOBAL_ATTENTION_LAYERS[0]

local_layer_idx = next(
    i for i in range(N_LAYER)
    if i not in GLOBAL_ATTENTION_LAYERS
)

global_attention = model.blocks[
    global_layer_idx
].attention

local_attention = model.blocks[
    local_layer_idx
].attention

test_length = LOCAL_ATTN_WINDOW + 44

global_mask = global_attention._build_attention_mask(
    query_length=test_length,
    key_length=test_length,
    query_position_offset=0,
    device=DEVICE
)

local_mask = local_attention._build_attention_mask(
    query_length=test_length,
    key_length=test_length,
    query_position_offset=0,
    device=DEVICE
)

last_position = test_length - 1

first_local_position = (
    last_position
    - LOCAL_ATTN_WINDOW
    + 1
)

print("Global layer:", global_layer_idx)
print("Local layer:", local_layer_idx)
print("Local window:", LOCAL_ATTN_WINDOW)
print("Test length:", test_length)

# ---------------------------------------------------------
# Global layer: full causal attention
# ---------------------------------------------------------

assert global_mask[
    last_position, 0
].item() is True

assert global_mask[
    last_position, last_position
].item() is True

# Future token must remain blocked.
assert global_mask[
    0, 1
].item() is False


# ---------------------------------------------------------
# Local layer: sliding-window causal attention
# ---------------------------------------------------------

# Token outside the local window must be blocked.
assert local_mask[
    last_position, 0
].item() is False

# First token inside the window must be visible.
assert local_mask[
    last_position,
    first_local_position
].item() is True

# Current token must be visible.
assert local_mask[
    last_position,
    last_position
].item() is True


# ---------------------------------------------------------
# Cached decoding at an absolute position
# ---------------------------------------------------------

past_length = 300
cached_key_length = past_length + 1

cached_global_mask = (
    global_attention._build_attention_mask(
        query_length=1,
        key_length=cached_key_length,
        query_position_offset=past_length,
        device=DEVICE
    )
)

cached_local_mask = (
    local_attention._build_attention_mask(
        query_length=1,
        key_length=cached_key_length,
        query_position_offset=past_length,
        device=DEVICE
    )
)

local_first_cached_key = (
    past_length
    - LOCAL_ATTN_WINDOW
    + 1
)

assert cached_global_mask.shape == (
    1,
    past_length + 1
)

assert cached_local_mask.shape == (
    1,
    past_length + 1
)

# Global layer can attend to the complete history.
assert cached_global_mask[
    0, 0
].item() is True

assert cached_global_mask[
    0, past_length
].item() is True

# Local layer blocks just-outside-window position.
assert cached_local_mask[
    0,
    local_first_cached_key - 1
].item() is False

# Local layer allows first position in the window.
assert cached_local_mask[
    0,
    local_first_cached_key
].item() is True

# Local layer allows the current token.
assert cached_local_mask[
    0,
    past_length
].item() is True

print("\nHybrid local/global attention mask check passed.")
```

```python
# KV Cache Equivalence Test
# During generation, KV caching should give the same results
# as processing the complete sequence normally.
# We compare:
# 1. Full-sequence forward pass
# 2. Token-by-token forward pass using the KV cache

model.eval()

torch.manual_seed(SEED)

cache_test_ids = torch.randint(
    low=0,
    high=tokenizer.vocab_size,
    size=(1, 16),
    device=DEVICE
)

with torch.no_grad():

    # Full sequence in one forward pass.
    full_logits, _ = model(
        cache_test_ids,
        use_cache=False
    )

    # Process one token at a time while keeping the
    # previous keys and values in the cache.
    past_key_values = None
    cached_logits = []

    for position in range(
        cache_test_ids.size(1)
    ):

        current_token = cache_test_ids[
            :, position:position + 1
        ]

        logits, past_key_values = model(
            current_token,
            past_key_values=past_key_values,
            use_cache=True
        )

        cached_logits.append(logits)

    cached_logits = torch.cat(
        cached_logits,
        dim=1
    )


# Compare full-sequence and cached outputs.
max_difference = (
    full_logits - cached_logits
).abs().max().item()

mean_difference = (
    full_logits - cached_logits
).abs().mean().item()

print(
    f"Maximum logit difference: {max_difference:.10f}"
)

print(
    f"Mean logit difference:    {mean_difference:.10f}"
)

assert torch.allclose(
    full_logits,
    cached_logits,
    atol=1e-4,
    rtol=1e-4
)

print("\nKV cache equivalence check passed.")
```

## 10. Training Setup and Optimization Hyperparameters
AdamW optimizer with decoupled weight decay (0.01), cosine learning rate scheduler, and gradient clipping (1.0).


```python
import os
import json
import math
import time

# Fresh optimizer for the freshly initialized model.
optimizer = torch.optim.AdamW(
    model.parameters(),
    lr=LEARNING_RATE,
    betas=(0.9, 0.95),
    weight_decay=WEIGHT_DECAY
)

# Number of tokens processed in one optimizer step.
TOKENS_PER_STEP = (
    PER_DEVICE_BATCH_SIZE
    * GRADIENT_ACCUMULATION_STEPS
    * TRAIN_BLOCK_SIZE
)

MAX_TRAIN_STEPS = (
    len(train_loader) // GRADIENT_ACCUMULATION_STEPS
)

print(f"Tokens per optimizer step: {TOKENS_PER_STEP:,}")
print(f"Total training tokens: {TRAIN_TOKENS:,}")
print(f"Planned optimizer steps: {MAX_TRAIN_STEPS:,}")

# Cosine learning-rate schedule with warmup.
def get_learning_rate(step):
    if step < WARMUP_STEPS:
        return LEARNING_RATE * (
            (step + 1) / WARMUP_STEPS
        )

    progress = (
        step - WARMUP_STEPS
    ) / max(
        1,
        MAX_TRAIN_STEPS - WARMUP_STEPS
    )

    progress = min(max(progress, 0.0), 1.0)

    return LEARNING_RATE * 0.5 * (
        1.0 + math.cos(math.pi * progress)
    )

print("Training setup ready.")
```

```python
# Checkpoint Setup

CHECKPOINT_DIR = os.path.join(
    OUTPUT_DIR,
    "checkpoints"
)

os.makedirs(
    CHECKPOINT_DIR,
    exist_ok=True
)

LATEST_CHECKPOINT = os.path.join(
    CHECKPOINT_DIR,
    "latest.pt"
)

BEST_CHECKPOINT = os.path.join(
    CHECKPOINT_DIR,
    "best.pt"
)

history = {
    "step": [],
    "train_loss": [],
    "val_loss": [],
    "val_perplexity": [],
    "learning_rate": [],
    "tokens_seen": []
}

best_val_loss = float("inf")

print("Checkpoint directory:", CHECKPOINT_DIR)
```

```python
# ---------------------------------------------------------
# 12. Validation
# ---------------------------------------------------------
# Evaluate the model without updating its weights.
# ---------------------------------------------------------

@torch.no_grad()
def evaluate_model():
    model.eval()

    total_loss = 0.0
    total_tokens = 0

    for batch in val_loader:

        input_ids = batch["input_ids"].to(
            DEVICE,
            non_blocking=True
        )

        labels = batch["labels"].to(
            DEVICE,
            non_blocking=True
        )

        logits, _ = model(
            input_ids,
            use_cache=False
        )

        loss = F.cross_entropy(
            logits.reshape(
                -1,
                tokenizer.vocab_size
            ),
            labels.reshape(-1)
        )

        tokens = labels.numel()

        total_loss += (
            loss.item() * tokens
        )

        total_tokens += tokens

    val_loss = (
        total_loss / total_tokens
    )

    val_perplexity = math.exp(
        min(val_loss, 20)
    )

    model.train()

    return val_loss, val_perplexity

print("Validation function ready.")
```

## 11. Pretraining Execution Loop
Autoregressive next-token prediction pretraining across token batches with cross-entropy loss tracking.


```python
# V1 training loop

model.train()

global_step = 0
tokens_seen = 0
start_time = time.time()

optimizer.zero_grad(set_to_none=True)

train_iterator = iter(train_loader)

while global_step < MAX_TRAIN_STEPS:

    accumulated_loss = 0.0

    for micro_step in range(GRADIENT_ACCUMULATION_STEPS):

        batch = next(train_iterator)

        input_ids = batch["input_ids"].to(
            DEVICE,
            non_blocking=True
        )

        labels = batch["labels"].to(
            DEVICE,
            non_blocking=True
        )

        logits, _ = model(
            input_ids,
            use_cache=False
        )

        loss = F.cross_entropy(
            logits.reshape(-1, tokenizer.vocab_size),
            labels.reshape(-1)
        )

        accumulated_loss += loss.item()

        loss = loss / GRADIENT_ACCUMULATION_STEPS
        loss.backward()

    current_lr = get_learning_rate(global_step)

    for param_group in optimizer.param_groups:
        param_group["lr"] = current_lr

    grad_norm = torch.nn.utils.clip_grad_norm_(
        model.parameters(),
        MAX_GRAD_NORM
    )

    optimizer.step()
    optimizer.zero_grad(set_to_none=True)

    global_step += 1
    tokens_seen += TOKENS_PER_STEP

    train_loss = (
        accumulated_loss / GRADIENT_ACCUMULATION_STEPS
    )

    if global_step == 1 or global_step % LOG_EVERY_STEPS == 0:

        elapsed = time.time() - start_time
        tokens_per_second = tokens_seen / max(elapsed, 1e-6)

        print(
            f"Step {global_step:,}/{MAX_TRAIN_STEPS:,} | "
            f"Loss: {train_loss:.4f} | "
            f"LR: {current_lr:.2e} | "
            f"Grad: {grad_norm.item():.3f} | "
            f"Tokens: {tokens_seen:,} | "
            f"Tok/s: {tokens_per_second:,.0f}"
        )

        history["step"].append(global_step)
        history["train_loss"].append(train_loss)
        history["learning_rate"].append(current_lr)
        history["tokens_seen"].append(tokens_seen)

    if (
        global_step % EVAL_EVERY_STEPS == 0
        or global_step == MAX_TRAIN_STEPS
    ):

        val_loss, val_ppl = evaluate_model()

        print(
            f"\nValidation | "
            f"Step {global_step:,} | "
            f"Loss: {val_loss:.4f} | "
            f"Perplexity: {val_ppl:.2f}\n"
        )

        history["val_loss"].append(val_loss)
        history["val_perplexity"].append(val_ppl)

        if val_loss < best_val_loss:

            best_val_loss = val_loss

            torch.save(
                {
                    "step": global_step,
                    "model_state_dict": model.state_dict(),
                    "optimizer_state_dict": optimizer.state_dict(),
                    "best_val_loss": best_val_loss,
                    "history": history
                },
                BEST_CHECKPOINT
            )

            print("New best checkpoint saved.")

    if (
        global_step % SAVE_EVERY_STEPS == 0
        or global_step == MAX_TRAIN_STEPS
    ):

        torch.save(
            {
                "step": global_step,
                "model_state_dict": model.state_dict(),
                "optimizer_state_dict": optimizer.state_dict(),
                "best_val_loss": best_val_loss,
                "history": history
            },
            LATEST_CHECKPOINT
        )

        with open(
            os.path.join(
                OUTPUT_DIR,
                "training_history.json"
            ),
            "w",
            encoding="utf-8"
        ) as f:
            json.dump(
                history,
                f,
                indent=2
            )

        print("Latest checkpoint saved.")

print("\nV1 training complete.")
print(f"Best validation loss: {best_val_loss:.6f}")
```

## 19. Autoregressive Text Generation
Greedy and nucleus sampling generation of legal continuations up to 128 words.


```python
# Generate legal-text continuations using Architecture 2
# Lightweight Hybrid Transformer and KV cache.

def apply_repetition_penalty(logits, generated_tokens, penalty):
    if penalty == 1.0 or len(generated_tokens) == 0:
        return logits

    for token_id in set(generated_tokens):
        if logits[token_id] < 0:
            logits[token_id] *= penalty
        else:
            logits[token_id] /= penalty

    return logits


def sample_next_token(logits):
    logits = logits / max(TEMPERATURE, 1e-6)

    logits[PAD_TOKEN_ID] = float("-inf")
    logits[BOS_TOKEN_ID] = float("-inf")

    if TOP_K > 0:
        k = min(TOP_K, logits.numel())

        top_values, _ = torch.topk(
            logits,
            k
        )

        cutoff = top_values[-1]
        logits[logits < cutoff] = float("-inf")

    if TOP_P < 1.0:
        sorted_logits, sorted_indices = torch.sort(
            logits,
            descending=True
        )

        sorted_probs = torch.softmax(
            sorted_logits,
            dim=-1
        )

        cumulative_probs = torch.cumsum(
            sorted_probs,
            dim=-1
        )

        remove = cumulative_probs > TOP_P

        remove[1:] = remove[:-1].clone()
        remove[0] = False

        sorted_logits[remove] = float("-inf")

        logits = torch.full_like(
            logits,
            float("-inf")
        )

        logits.scatter_(
            0,
            sorted_indices,
            sorted_logits
        )

    probabilities = torch.softmax(
        logits,
        dim=-1
    )

    return torch.multinomial(
        probabilities,
        num_samples=1
    ).item()


def generate_one(
    prompt,
    model,
    tokenizer,
    device,
    target_words=128
):
    token_ids = tokenizer.encode(prompt)

    max_prompt_tokens = (
        MODEL_CONTEXT_LENGTH - MAX_GENERATION_TOKENS
    )

    # Keep the most recent part of the prompt if it is too long.
    if len(token_ids) > max_prompt_tokens:
        token_ids = token_ids[-max_prompt_tokens:]

    input_ids = torch.tensor(
        [token_ids],
        dtype=torch.long,
        device=device
    )

    generated_tokens = []

    with torch.inference_mode():

        # First pass processes the complete prompt and builds the KV cache.
        logits, past_key_values = model(
            input_ids,
            use_cache=True
        )

        next_logits = logits[:, -1, :][0]

        while len(generated_tokens) < MAX_GENERATION_TOKENS:

            next_logits = apply_repetition_penalty(
                next_logits,
                generated_tokens,
                REPETITION_PENALTY
            )

            next_token = sample_next_token(
                next_logits.clone()
            )

            generated_tokens.append(
                next_token
            )

            generated_text = tokenizer.decode(
                generated_tokens
            )

            if len(generated_text.split()) >= target_words:
                break

            current_length = (
                len(token_ids)
                + len(generated_tokens)
            )

            if current_length >= MODEL_CONTEXT_LENGTH:
                break

            next_input = torch.tensor(
                [[next_token]],
                dtype=torch.long,
                device=device
            )

            logits, past_key_values = model(
                next_input,
                past_key_values=past_key_values,
                use_cache=True
            )

            next_logits = logits[:, -1, :][0]

    generated_text = tokenizer.decode(
        generated_tokens
    )

    generated_words = generated_text.split()

    return " ".join(
        generated_words[:target_words]
    )


def generate_batch(
    prompts,
    model,
    tokenizer,
    device,
    target_words=128
):
    predictions = []

    for prompt in prompts:
        prediction = generate_one(
            prompt=prompt,
            model=model,
            tokenizer=tokenizer,
            device=device,
            target_words=target_words
        )

        predictions.append(prediction)

    return predictions


print("Architecture 2 generation functions ready.")
print("Generation uses the KV cache.")
print("Hybrid global/local attention is active.")
```
